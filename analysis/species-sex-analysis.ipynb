```python
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

# Go one directory up from analysis/
BASE_DIR = Path.cwd().parent

DATA_FILE = BASE_DIR / "data" / "penguins.csv"
FIGURES_DIR = BASE_DIR / "figures"

# Create figures folder if necessary
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

print("Project directory:", BASE_DIR)
print("Dataset found:", DATA_FILE.exists())
# Load the dataset
df = pd.read_csv(DATA_FILE)

# Remove observations with missing sex
df_clean = df.dropna(subset=["species", "sex"]).copy()

# Count males and females for each species
species_sex = pd.crosstab(
    df_clean["species"],
    df_clean["sex"]
)

print("\n=== SPECIES BY SEX ===")
print(species_sex)

# Calculate percentages within each species
percentages = species_sex.div(
    species_sex.sum(axis=1), axis=0
) * 100

print("\n=== PERCENTAGE BY SPECIES ===")
print(percentages.round(2))

# Plot counts
ax = species_sex.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Penguin Sex Distribution by Species")
plt.xlabel("Penguin Species")
plt.ylabel("Number of Penguins")
plt.xticks(rotation=0)
plt.legend(title="Sex")
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "species_sex_distribution.png",
    dpi=150
)
plt.close()

# Plot percentages
percentages.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5)
)

plt.title("Male and Female Percentage by Species")
plt.xlabel("Penguin Species")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.legend(title="Sex")
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "species_sex_percentage.png",
    dpi=150
)
plt.close()

print("\nAnalysis completed successfully!")
```
